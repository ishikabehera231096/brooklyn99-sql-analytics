```python
import pandas as pd 
```

Load the merged dataset

```python
df = pd.read_csv("/Users/ishik/Desktop/brooklyn99-sql-analytics/exports/b99_episodes_with_writer_director.csv")
df.head()
```

```python
df.columns
```

Average rating and total episodes per first credited director

```python
director_stats = df.groupby("Director_0").agg(avg_rating =("Rating","mean"),total_episodes=("Rating","count")).sort_values("avg_rating",ascending=False)
director_stats
```

Consider the first credited directors that have directed more than 3 episodes

```python
director_stats[director_stats["total_episodes"] > 3].sort_values("total_episodes",ascending=False)
```

Average rating and episodes per first credited writer

```python
writer_stats = df.groupby("Writer_0").agg(avg_rating = ("Rating","mean"),total_episodes=("Rating","count")).sort_values("total_episodes",ascending=False)
writer_stats
```

Count the total episodes directed by directors across all credits

```python
all_directors = pd.concat([df['Director_0'], df['Director_1']], ignore_index=True).dropna()
all_directors.value_counts()
```

Count the total episodes writen by writers across all credits

```python
all_writers = pd.concat((df["Writer_0"],df["Writer_1"],df["Writer_2"]),ignore_index=True).dropna()
all_writers.value_counts()
```
